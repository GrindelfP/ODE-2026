## Задача 1. Бак с рассолом

В бак объёмом $V = 100$ л налита чистая вода. В бак непрерывно поступает раствор соли с концентрацией $c_{in} = 0{,}5$ кг/л со скоростью $q = 5$ л/мин. Смесь идеально перемешивается и вытекает с той же скоростью $q$, поэтому объём остаётся постоянным.

**Найти** массу соли $m(t)$ в баке и построить её график.

**Вывод ДУ.** Скорость изменения массы соли = приток − отток:

$$m' = c_{in}\,q - \frac{m}{V}\,q
\quad\Longrightarrow\quad
m' + \frac{q}{V}\,m = c_{in}\,q,\qquad m(0)=0.$$

Это ЛНДУ1П с постоянными коэффициентами.

## Задача 2. Температура корпуса спутника

Спутник обращается вокруг Земли с периодом $P = 90$ мин. Из них $t_{sh} = 35$ мин он проводит в тени Земли, остальные $t_{sun} = 55$ мин - на свету. Тепловой обмен с окружением линеаризуем: корпус стремится к равновесной температуре $T_{eq}$ с постоянной времени $\tau = 30$ мин:

$$T' = -\frac{T - T_{eq}}{\tau}
\quad\Longrightarrow\quad
T' + \frac{1}{\tau}\,T = \frac{T_{eq}}{\tau}.$$

Равновесная температура **кусочно-постоянна**: $T_{eq}=T_{sun}=+80\,°C$ на свету и $T_{eq}=T_{sh}=-40\,°C$ в тени.
В момент $t=0$ спутник выходит из тени на свет с $T=T_0=-20\,°C$.

**Найти** $T(t)$ за 5 витков и построить график.

### Как разбить задачу

Единое $T_{eq}(t)$ с разрывами подавать в `dsolve` не нужно. На каждом участке $T_{eq}$ - **константа**, поэтому получаем два простых ЛНДУ1П:

| Участок | Длительность | Уравнение |
|---|---|---|
| Свет | $t_{sun}$ | $T' + T/\tau = T_{sun}/\tau$ |
| Тень | $t_{sh}$ | $T' + T/\tau = T_{sh}/\tau$ |

На каждом участке вводим **локальное время** $s\in[0,\,\text{длительность}]$ и решаем ДУ с произвольным начальным значением $T_{start}$:
$T(0)=T_{start}$. Решение получается один раз, потом переиспользуется для любого $T_{start}$.

### Как сшить

Температура корпуса - непрерывная функция (тепловая инерция), поэтому:

1. Конец участка = начало следующего: $T_{start}^{(n+1)} = T^{(n)}(\text{конец участка } n)$.
2. Глобальное время: $t = t_{начала\ участка} + s$.
3. Порядок: свет → тень → свет → тень ... на каждом витке.

Так сшивка сводится к подстановке числа в начальное условие.
