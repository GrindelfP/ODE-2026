## Модель «хищник — жертва»

Пусть $t$ — время, $x(t)$ — численность жертв, а $y(t)$ — численность хищников. Считаем все коэффициенты модели положительными и изучаем неотрицательные численности.

Предположим, что при отсутствии хищников жертвы размножаются со скоростью, пропорциональной их численности: вклад в изменение $x$ равен $ax$, где $a>0$. При отсутствии жертв хищникам не хватает пищи, и их численность убывает со скоростью, пропорциональной $y$: вклад в изменение $y$ равен $-cy$, где $c>0$.

Пусть популяции перемешаны, а частота встреч пропорциональна числу возможных пар «жертва — хищник», то есть произведению $xy$. Каждая встреча уменьшает численность жертв и способствует увеличению численности хищников. Поэтому вклады встреч в скорости изменения равны соответственно $-bxy$ и $+dxy$, где $b,d>0$. Коэффициент $b$ описывает потери жертв при встречах, а $d$ — прирост хищников благодаря добыче; эти коэффициенты не обязаны совпадать.

Складывая вклады без взаимодействия и вклады встреч, получаем общую систему Лотки — Вольтерры:
$$x'=ax-bxy,\qquad y'=-cy+dxy.$$

Далее положим $a=2$, $b=1$, $c=1$, $d=1$. Тогда исследуем конкретную систему
$$x'=x(2-y),\qquad y'=y(x-1).$$
